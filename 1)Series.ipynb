# Pandas Series

##### A Series is a one-dimensional labeled array capable of holding any data type. The axis labels are collectively called the index.

# Creating Series

##### There are multiple ways to create a pandas Series:

```python
import pandas as pd
import numpy as np
```

```python
lebels = ['a','b','c']
my_list = [10,20,30]
arr = np.array([10,20,30])
d1 = {1:10,2:20,3:30}
d2 = {'a':10,'b':20,'c':30}
```

```python
pd.Series(my_list)
```

```python
pd.Series(my_list,index = lebels)
```

```python
pd.Series(arr,index=lebels)
```

```python
pd.Series(d1)
```

```python
pd.Series(d2)
```
